```python
import shutil, os

source_dir = "/workspaces/zef-/game logos"
dest_dir = "/workspaces/zef-/public/games"
os.makedirs(dest_dir, exist_ok=True)

file_mapping = {
    "BGMI.jpeg": "bgmi.jpeg",
    "clash of clans.png": "clash-of-clans.png",
    "clash royale.png": "clash-royale.png",
    "free fire max.png": "free-fire-max.png",
    "league of legends.png": "league-of-legends.png",
    "pokemon unite.png": "pokemon-unite.png",
    "pubg mobile.png": "pubg-mobile.png",
    "valorant.png": "valorant.png",
}

for src_name, dst_name in file_mapping.items():
    src = os.path.join(source_dir, src_name)
    dst = os.path.join(dest_dir, dst_name)
    shutil.copy2(src, dst)
    size = os.path.getsize(dst)
    print(f"OK: {src_name} -> {dst_name} ({size:,} bytes)")

print("\nFiles in destination:")
for f in sorted(os.listdir(dest_dir)):
    fp = os.path.join(dest_dir, f)
    print(f"  {f} ({os.path.getsize(fp):,} bytes)")
```
